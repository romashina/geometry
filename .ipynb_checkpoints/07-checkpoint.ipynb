# Геометрия

**Геометрия** (*греч.*: **гео** - земля, **метрео** - мерить) - землемерие. Наука, занимающаяся изучением фигур. Она состоит:

- **Планиметрия** - фигуры на плоскости
- **Стереометрия** - фигуры в пространстве

## 1. Прямая и отрезок
