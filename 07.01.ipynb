10. Начертите пятиугольник, как на рисунке 9.

- a. Разбейте его на треугольники так, чтобы у пятиугольника и каждого треугольника были общие стороны. 
- b. Сколько треугольников может получиться на чертеже?
- c. Какие ещё многоугольники получились на чертеже?

```python
# # Координаты середины отрезка EB
# E = (20, 140)
# B = (180, 120)
# C = (130, 200)
# # Считаем середину через генератор или распаковку
# print(f"F{tuple((a + b) / 2 for a, b in zip(E, B))}")
# print(f"F{tuple((a + b) / 2 for a, b in zip(E, C))}")
def bias(a: tuple[int, int], b: tuple[int, int], t: float) -> tuple[float, float]:
    """Возвращает координаты точки на отрезке AB со смещением t относительно точки B
    :param a: координаты точки A в виде кортежа (x, y)
    :param b: координаты точки B в виде кортежа (x, y)
    :param t: смещение точки B ближе к A (0 <= t <= 1), например 0,9 - 90% отрезка AB
              отдаление точки В в от точки A (t > 1) 
    :return: координаты точки F в виде кортежа (x, y)
    """
    return (a[0] +(b[0] - a[0]) * t, a[1] +(b[1] - a[1]) * t)

bias((20, 140), (180, 120), 0.5)
```

- a.

- b. 5 треугольников в одном пятиугольнике
- c. 4-угольник, 6-угольник, 7-угольник

24. Луч ОС делит угол AOB на два угла. Сравните углы AOB и АОС.

```python
import numpy as np
import matplotlib.pyplot as plt

plt.figure(figsize=(5, 5))

B = np.r_[0, 3]
O = np.r_[0, 0]
A = np.r_[3, 0] 
C = (A + B)/2

points = {
    'B': B,
    'O': O,
    'A': A,
    'C': C,
}

plt.plot(*zip(O, A), 'k-')
plt.plot(*zip(O, C), 'g-')
plt.plot(*zip(O, B), 'r-')

plt.plot(*zip(*points.values()), 'ko')
for name, coord in points.items():
    plt.text(*(coord + [0.1, 0.1]), name, fontsize=12, va='bottom', ha='left')


# Дуга от 0 до 45 градусов (в радианах)
AOC = np.linspace(0, np.radians(45), 50)
AOB = np.linspace(0, np.radians(90), 100)
# X = r * cos(a), Y = r * sin(a)
plt.plot(0.5 * np.cos(AOC), 0.5 * np.sin(AOC), 'g-')
plt.plot(np.cos(AOB), np.sin(AOB), 'r-')

plt.axis('off')
plt.show()
```

**Ответ**: $\angle AOB > \angle AOC$

25. Луч l — биссектриса угла hk. Можно ли наложением совместить углы:

- а) hl и lk;
- б) hl и hk?

```python
import numpy as np
import matplotlib.pyplot as plt

plt.figure(figsize=(5, 5))

H = np.r_[0, 3]
O = np.r_[0, 0]
K = np.r_[3, 0] 
L = ((H + K)/2)+0.7

points = {
    'h': H,
    'k': K,
    'l': L,
}

plt.plot(*zip(O, H), 'k-')
plt.plot(*zip(O, L), 'g-')
plt.plot(*zip(O, K), 'r-')

# plt.plot(*zip(*points.values()), 'ko')
for name, coord in points.items():
    plt.text(*(coord + [0.1, 0.1]), name, fontsize=12, va='bottom', ha='left')


# Дуга от 0 до 45 градусов (в радианах)
hl = np.linspace(0, np.radians(45), 50)
hk = np.linspace(0, np.radians(90), 100)
# X = r * cos(a), Y = r * sin(a)
plt.plot(0.5 * np.cos(hl), 0.5 * np.sin(hl), 'g-')
plt.plot(np.cos(hk), np.sin(hk), 'r-')

plt.axis('off')
plt.show()
```

**Ответы**:

- а) Да, можно, т.к.: $\angle hl = \angle lk$
- б) Нет, нельзя, т.к.: $\angle hl \ne \angle hk$
