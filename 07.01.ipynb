10. Начертите пятиугольник, как на рисунке 9.

- a. Разбейте его на треугольники так, чтобы у пятиугольника и каждого треугольника были общие стороны. 
- b. Сколько треугольников может получиться на чертеже?
- c. Какие ещё многоугольники получились на чертеже?

```python
# # Координаты середины отрезка EB
# E = (20, 140)
# B = (180, 120)
# C = (130, 200)
# # Считаем середину через генератор или распаковку
# print(f"F{tuple((a + b) / 2 for a, b in zip(E, B))}")
# print(f"F{tuple((a + b) / 2 for a, b in zip(E, C))}")
def bias(a: tuple[int, int], b: tuple[int, int], t: float) -> tuple[float, float]:
    """Возвращает координаты точки на отрезке AB со смещением t относительно точки B
    :param a: координаты точки A в виде кортежа (x, y)
    :param b: координаты точки B в виде кортежа (x, y)
    :param t: смещение точки B ближе к A (0 <= t <= 1), например 0,9 - 90% отрезка AB
              отдаление точки В в от точки A (t > 1) 
    :return: координаты точки F в виде кортежа (x, y)
    """
    return (a[0] +(b[0] - a[0]) * t, a[1] +(b[1] - a[1]) * t)

bias((20, 140), (180, 120), 0.5)
```

- b. 5 треугольников в одном пятиугольнике
- c. 4-угольник, 6-угольник, 7-угольник

24. Луч ОС делит угол AOB на два угла. Сравните углы AOB и АОС.

```python
import numpy as np
import matplotlib.pyplot as plt

plt.figure(figsize=(5, 5))

B = np.r_[0, 3]
O = np.r_[0, 0]
A = np.r_[3, 0] 
C = (A + B)/2

points = {
    'B': B,
    'O': O,
    'A': A,
    'C': C,
}

plt.plot(*zip(O, A), 'k-')
plt.plot(*zip(O, C), 'g-')
plt.plot(*zip(O, B), 'r-')

plt.plot(*zip(*points.values()), 'ko')
for name, coord in points.items():
    plt.text(*(coord + [0.1, 0.1]), name, fontsize=12, va='bottom', ha='left')


# Дуга от 0 до 45 градусов (в радианах)
AOC = np.linspace(0, np.radians(45), 50)
AOB = np.linspace(0, np.radians(90), 100)
# X = r * cos(a), Y = r * sin(a)
plt.plot(0.5 * np.cos(AOC), 0.5 * np.sin(AOC), 'g-')
plt.plot(np.cos(AOB), np.sin(AOB), 'r-')

plt.axis('off')
plt.show()
```

**Ответ**: $\angle AOB > \angle AOC$

48. Начертите угол AOB и с помощью транспортира проведите луч ОС так, чтобы луч ОА являлся биссектрисой угла ВОС. Всегда ли это выполнимо?

```python
import numpy as np
import matplotlib.pyplot as plt

def draw_line(start: np.ndarray, length: float, angle_deg: float, format: str) -> np.ndarray:
    """
    Описание

    Функция рисует линию (вектор, отрезок, луч, прямую)
    нужного формата и наклона на координатной плоскости.
    
    Внешние параметры

    - start: координаты начальной точки
    - length: скалярная длина линии
    - angle_deg: угол наклона линии в градусах относительно оси x
    - format: строка формата matplotlib для линии (цвет, стиль линии, маркеры на концах)
        Например: 'k.-', 'b--', 'ro-'.

    Внутренние переменные

    - a: скаляр угла наклона линии в радианах
    - arc: numpy массив точек для построения дуги, генерируется с помощью функции np.linspace(start, a, angle_deg), 
        принимающей начальную точку линии, конечную точку (радианный скаляр) и количество точек для построения дуги
    - end: numpy массив с координатами конечной точки, где координата x вычисляется как начальная точка на оси x (start[0])
        прибавить проекцию длины линии на ось x (length * np.cos(a)), где проекция вычисляется как длина линии (length) 
        умноженная на косинус угла наклона линии (np.cos(a)), 
        а координата y вычисляется как начальная точка на оси y (start[1]) прибавить проекцию длины линии на ось y (length * np.sin(a)),
        где проекция вычисляется как длина линии (length) умноженная на синус угла наклона линии (np.sin(a)). 

    Использование:

    draw_line([x, y], length, angle_deg, format)
    """
    a = np.deg2rad(angle_deg)
    arc = np.linspace(start, a, int(angle_deg))
    plt.plot(start[0] + np.cos(arc)*angle_deg/100, start[1] + np.sin(arc)*angle_deg/100, f'{format[0]}--')
    end = np.array([start[0] + length * np.cos(a), start[1] + length * np.sin(a)])
    plt.plot(*zip(start, end), format)
    return end
```

```python
plt.figure(figsize=(5, 5))
plt.axis('equal')
plt.axis('off')
O = np.array([0, 0])
plt.text(*O-.2, 'O')
A = draw_line(O, 5, 20, 'k.-')
plt.text(*A+.1, 'A')
B = draw_line(O, 5, 40, 'b.-')
plt.text(*B+.1, 'B')
C = draw_line(O, 5, 0, 'g-')
plt.text(*C+0.1, 'C')

plt.show()
```

**Ответ**:
 
- (Для школьного курса, где транспортир имеет максимум 180°): Нет. Построение выполнимо, когда ∠АОВ острый или прямой. 
- (Для института где градусы могут быть и в несколько оборотов вокруг оси, например 1800°): Да, такое всегда выполнимо.

52. Луч ОС делит угол AOB на два угла. Найдите угол СОВ, если ∠AOB = 78°, а угол АОС на 18° меньше угла ВОС.

```python
import sympy as sp

def lin(equation):
    string = ""
    for i in range(len(equation)):
        if equation[i].isalpha() and equation[i-1].isdigit(): 
            string+="*"+equation[i]
        else:
            string+=equation[i]

    left, right = string.split("=")
    return " ".join([str(float(x)) if x.is_Float else str(x) for x in sp.solve(f"{left} -({right})")])
```

```python
AOB = 78
AOC = "BOC-18"
print("Ответ: угол BOC =", lin(f"{AOB} = {AOC}+BOC"), "градусов")
```
