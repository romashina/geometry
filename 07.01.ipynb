10. Начертите пятиугольник, как на рисунке 9.

- a. Разбейте его на треугольники так, чтобы у пятиугольника и каждого треугольника были общие стороны. 
- b. Сколько треугольников может получиться на чертеже?
- c. Какие ещё многоугольники получились на чертеже?

```python
# # Координаты середины отрезка EB
# E = (20, 140)
# B = (180, 120)
# C = (130, 200)
# # Считаем середину через генератор или распаковку
# print(f"F{tuple((a + b) / 2 for a, b in zip(E, B))}")
# print(f"F{tuple((a + b) / 2 for a, b in zip(E, C))}")
def bias(a: tuple[int, int], b: tuple[int, int], t: float) -> tuple[float, float]:
    """Возвращает координаты точки на отрезке AB со смещением t относительно точки B
    :param a: координаты точки A в виде кортежа (x, y)
    :param b: координаты точки B в виде кортежа (x, y)
    :param t: смещение точки B ближе к A (0 <= t <= 1), например 0,9 - 90% отрезка AB
              отдаление точки В в от точки A (t > 1) 
    :return: координаты точки F в виде кортежа (x, y)
    """
    return (a[0] +(b[0] - a[0]) * t, a[1] +(b[1] - a[1]) * t)

bias((20, 140), (180, 120), 0.5)
```

- b. 5 треугольников в одном пятиугольнике

21. На луче с началом О отмечены точки A, В и С так, что точ­ка В лежит между точками О и А, а точка А — между точка­ми О и С. Сравните отрезки ОВ и ОА, ОС и ОА, ОВ и ОС.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 1))
points = {
    'O': (0, 0),
    'B': (3, 0),
    'A': (6, 0),
    'C': (9, 0)
}

plt.plot(*zip((0, 0), (10,0)), 'k')
plt.plot(*zip(points['O'], points['B']), 'g')
plt.plot(*zip(points['B'], points['A']), 'y')
plt.plot(*zip(points['A'], points['C']), 'b')

plt.plot(*zip(*points.values()), 'ko')
for name, coord in points.items():
    plt.text(*coord, name, va='bottom', ha='right')
plt.axis('off')
plt.show()
```

**Ответ**: ОВ < ОА < OC
